```python

import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python

data = {
    'Student_ID': [101, 102, 103, 104, 105, 102, 106, 107],
    'Name': ['Anu', 'Ravi', 'Meera', 'JOHN', 'Asha', 'Ravi', 'Kiran', 'Diya'],
    'Age': [20, 21, np.nan, 22, 20, 21, np.nan, 23],
    'Marks': [85, 90, 78, np.nan, 88, 90, 75, 92],
    'Date': ['2026-01-10', '10/02/2026', '2026-03-15',
             '15-04-2026', '2026-05-20', '10/02/2026',
             '2026-06-12', '12/07/2026']
}

df = pd.DataFrame(data)

print(df)
```

```python

print("Dataset Information:")
print(df.info())

print("\nDataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())
```

```python

print("Number of duplicate rows:", df.duplicated().sum())

print("\nDuplicate Records:")
print(df[df.duplicated()])
```

```python

df = df.drop_duplicates()

print("Duplicates removed successfully!")
print(df)
```

```python

df['Age'] = df['Age'].fillna(df['Age'].median())

df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

print(df.isnull().sum())
```

```python

df['Name'] = df['Name'].str.strip().str.title()

print(df['Name'])
```

```python

df['Date'] = pd.to_datetime(
    df['Date'],
    format='mixed',
    dayfirst=True
)

print(df['Date'])
```

```python

print("Final Dataset:")
print(df)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python

df.to_csv('cleaned_student_data.csv', index=False)

print("Cleaned dataset saved successfully!")
```
